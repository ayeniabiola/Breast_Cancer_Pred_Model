```python
# First, we import our libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.svm import SVC
```

```python
can = pd.read_csv("data/breast_cancer_data.csv")
can
```

```python
can.isnull().sum()
```

```python
can["diagnosis"].value_counts()
```

```python
x = can.drop(columns=['id', 'diagnosis', 'Unnamed: 32'])
y = can['diagnosis']

# x = can.drop(columns=['id', 'diagnosis','Unnamed: 32'])
# y = can['diagnosis']
```

```python
x.head()
```

```python
y.head()
```

```python
y= y.map({"M": 0, "B": 1 })
```

```python
y.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split (x, y, test_size = 0.2, random_state = 42, stratify = y)
```

```python
print("x_train:", x_train.shape)
print("x_test:", x_test.shape)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

# USING LOGISTIC REGRESSION

```python
log_model = LogisticRegression()
log_model.fit(x_train_scaled, y_train)
```

```python
log_prediction = log_model.predict(x_test_scaled)

log_accuracy = accuracy_score(
    y_test,
    log_prediction
)
print("Logistic Regression:", log_accuracy)
```

```python
print(classification_report(y_test, log_prediction))
```

## USING PIPELINE

```python
from sklearn.pipeline import Pipeline
pipeline = Pipeline([
 ('scaler', StandardScaler()),
 ('model', LogisticRegression())
])
pipeline.fit(x_train, y_train)
```

```python
log_pred = pipeline.predict(x_test_scaled)
log_accuracy = accuracy_score(
    y_test,
    log_pred
)
print("Logistic Regression:", log_accuracy)
```

# USING DECISION TREE

```python
tree_model = DecisionTreeClassifier(max_depth =3, random_state=42)
tree_model.fit(x_train_scaled, y_train)                                    
```

```python
tree_pred = tree_model.predict(x_test_scaled)
tree_accuracy = accuracy_score(
    y_test,
    tree_pred
)
print("Decision Tree Classifier:", tree_accuracy)
```

```python
print(classification_report(y_test,tree_pred))
```

```python
pipeline2 = Pipeline([
    ('scaler', StandardScaler()),
    ('model', DecisionTreeClassifier())
])
pipeline2.fit(x_train,y_train)
```

```python
tree_pred = pipeline2.predict(x_test_scaled)
tree_accuracy = accuracy_score(
    y_test,
    tree_pred
)
print("Decision Tree Classifier:", tree_accuracy)
```
